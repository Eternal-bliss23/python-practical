## Q1. Student Result Processing System

Enter name, roll number and marks of 5 subjects. Find total, percentage,
average, grade, pass/fail in each subject and print a marksheet.

```python
name = input("Enter student name: ")
roll = input("Enter roll number: ")

marks = []
for i in range(5):
    mark = float(input("Enter marks of subject " + str(i + 1) + ": "))
    marks.append(mark)

total = sum(marks)
percentage = total / 5
average = total / 5

if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B+"
elif percentage >= 60:
    grade = "B"
elif percentage >= 50:
    grade = "C"
elif percentage >= 40:
    grade = "D"
else:
    grade = "F"

print()
print("------ MARKSHEET ------")
print("Name        :", name)
print("Roll Number :", roll)
print()
print("Subject wise result:")

failed = 0
for i in range(5):
    if marks[i] >= 40:
        print("  Subject", i + 1, ":", marks[i], "- Pass")
    else:
        print("  Subject", i + 1, ":", marks[i], "- Fail")
        failed = failed + 1

print()
print("Total marks :", total)
print("Percentage  :", round(percentage, 2), "%")
print("Average     :", round(average, 2))
print("Grade       :", grade)

if failed == 0:
    print("Result      : PASS")
else:
    print("Result      : FAIL")
```

## Q3. String Analysis System

Count vowels, consonants, digits and special characters, reverse the string,
check palindrome, count words and print the frequency of each character.

```python
sentence = input("Enter a sentence: ")

vowels = 0
consonants = 0
digits = 0
special = 0

for ch in sentence:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels = vowels + 1
        else:
            consonants = consonants + 1
    elif ch.isdigit():
        digits = digits + 1
    elif ch != " ":
        special = special + 1

reverse = sentence[::-1]

clean = ""
for ch in sentence.lower():
    if ch.isalnum():
        clean = clean + ch

if clean == clean[::-1]:
    palindrome = "Yes"
else:
    palindrome = "No"

words = sentence.split()

frequency = {}
for ch in sentence.lower():
    if ch != " ":
        if ch in frequency:
            frequency[ch] = frequency[ch] + 1
        else:
            frequency[ch] = 1

print()
print("Original string    :", sentence)
print("Reverse string     :", reverse)
print("Vowels             :", vowels)
print("Consonants         :", consonants)
print("Digits             :", digits)
print("Special characters :", special)
print("Number of words    :", len(words))
print("Palindrome         :", palindrome)
print()
print("Character frequency:")
for ch in sorted(frequency):
    print(" ", ch, "=", frequency[ch])
```

## Q5. Number Analysis Using Loops

Prime, Armstrong, palindrome, factorial and multiplication table.

```python
num = int(input("Enter a number: "))

# prime
prime = True
if num <= 1:
    prime = False
else:
    i = 2
    while i < num:
        if num % i == 0:
            prime = False
            break
        i = i + 1

# armstrong
text = str(num)
power = len(text)
total = 0
for d in text:
    total = total + int(d) ** power
armstrong = (total == num)

# palindrome
palindrome = (text == text[::-1])

# factorial
fact = 1
for i in range(1, num + 1):
    fact = fact * i

print()
print("Number            :", num)
if prime:
    print("Prime number      : Yes")
else:
    print("Prime number      : No")
if armstrong:
    print("Armstrong number  : Yes")
else:
    print("Armstrong number  : No")
if palindrome:
    print("Palindrome number : Yes")
else:
    print("Palindrome number : No")
print("Factorial         :", fact)
print()
print("Multiplication table:")
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

## Q6. Shopping Cart Using Lists

Products and prices are kept in two lists. The menu adds, removes and searches
products, and shows the bill.

```python
products = ["Notebook", "Pen Drive", "Headphones"]
prices = [45, 550, 1200]

while True:
    print()
    print("1. Add product")
    print("2. Remove product")
    print("3. Search product")
    print("4. Show bill")
    print("5. Show all products")
    print("0. Exit")
    choice = input("Enter your choice: ")

    if choice == "1":
        product = input("Enter product name: ")
        price = float(input("Enter product price: "))
        products.append(product)
        prices.append(price)
        print(product, "added to cart")

    elif choice == "2":
        product = input("Enter product name to remove: ")
        if product in products:
            position = products.index(product)
            products.pop(position)
            prices.pop(position)
            print(product, "removed from cart")
        else:
            print("Product not found")

    elif choice == "3":
        product = input("Enter product name to search: ")
        if product in products:
            print("Product found. Price =", prices[products.index(product)])
        else:
            print("Product not found")

    elif choice == "4":
        total = 0
        print()
        print("------ BILL ------")
        for i in range(len(products)):
            print(products[i], "-", prices[i])
            total = total + prices[i]
        print("Total amount =", total)

    elif choice == "5":
        print()
        print("Products in cart:")
        for i in range(len(products)):
            print(i + 1, products[i], "-", prices[i])

    elif choice == "0":
        print("Thank you")
        break

    else:
        print("Wrong choice")
```

## Q11. Banking System Using User-Defined Functions

```python
account = {}

def create_account():
    name = input("Enter customer name: ")
    number = input("Enter account number: ")
    balance = float(input("Enter opening balance: "))
    account["name"] = name
    account["number"] = number
    account["balance"] = balance
    print("Account created for", name)

def deposit(amount):
    account["balance"] = account["balance"] + amount
    print("Amount deposited:", amount)
    print("Balance:", account["balance"])

def withdraw(amount):
    if amount > account["balance"]:
        print("Not enough balance. Withdrawal not allowed")
    else:
        account["balance"] = account["balance"] - amount
        print("Amount withdrawn:", amount)
        print("Balance:", account["balance"])

def check_balance():
    print("Account holder:", account["name"])
    print("Account number:", account["number"])
    print("Balance:", account["balance"])

create_account()

amount = float(input("Enter amount to deposit: "))
deposit(amount)

amount = float(input("Enter amount to withdraw: "))
withdraw(amount)

amount = float(input("Enter amount to withdraw again: "))
withdraw(amount)

print()
check_balance()
```

## Q12. Function-Based Number Operations

One function for each operation and a menu to call them.

```python
def is_prime(num):
    if num <= 1:
        return False
    i = 2
    while i < num:
        if num % i == 0:
            return False
        i = i + 1
    return True

def is_armstrong(num):
    text = str(num)
    power = len(text)
    total = 0
    for d in text:
        total = total + int(d) ** power
    return total == num

def factorial(num):
    fact = 1
    for i in range(1, num + 1):
        fact = fact * i
    return fact

def reverse_number(num):
    return int(str(num)[::-1])

def is_palindrome(num):
    return str(num) == str(num)[::-1]

while True:
    print()
    print("1. Prime number")
    print("2. Armstrong number")
    print("3. Factorial")
    print("4. Reverse of a number")
    print("5. Palindrome")
    print("0. Exit")
    choice = input("Enter your choice: ")

    if choice == "0":
        print("Program ended")
        break

    num = int(input("Enter a number: "))

    if choice == "1":
        if is_prime(num):
            print(num, "is a prime number")
        else:
            print(num, "is not a prime number")
    elif choice == "2":
        if is_armstrong(num):
            print(num, "is an Armstrong number")
        else:
            print(num, "is not an Armstrong number")
    elif choice == "3":
        print("Factorial of", num, "=", factorial(num))
    elif choice == "4":
        print("Reverse of", num, "=", reverse_number(num))
    elif choice == "5":
        if is_palindrome(num):
            print(num, "is a palindrome")
        else:
            print(num, "is not a palindrome")
    else:
        print("Wrong choice")
```

## Q13. Lambda, Map, Filter and Reduce

```python
from functools import reduce

numbers = [12, 7, 45, 88, 23, 56, 91, 34, 60, 15]
print("Numbers:", numbers)

square = lambda x: x * x
print()
print("Squares using lambda:")
for n in numbers:
    print(square(n), end=" ")
print()

cubes = list(map(lambda x: x ** 3, numbers))
print("Cubes using map     :", cubes)

even = list(filter(lambda x: x % 2 == 0, numbers))
print("Even numbers        :", even)

big = list(filter(lambda x: x > 50, numbers))
print("Numbers above 50    :", big)

total = reduce(lambda a, b: a + b, numbers)
product = reduce(lambda a, b: a * b, numbers)
print("Sum using reduce    :", total)
print("Product using reduce:", product)
```

## Q15. File Analysis and Word Frequency

```python
text = """Python is a popular programming language. It was created by Guido van Rossum
and released in 1991. Python is simple and easy to learn. It supports object
oriented programming. Python is used in data science and web development.
Practice is the only way to learn programming."""

file = open("paragraph.txt", "w")
file.write(text)
file.close()

file = open("paragraph.txt", "r")
content = file.read()
file.close()

lines = content.split("\n")
words = content.split()

print("Number of lines      :", len(lines))
print("Number of words      :", len(words))
print("Number of characters :", len(content))

count = {}
for word in words:
    word = word.lower().strip(".,!?")
    if word in count:
        count[word] = count[word] + 1
    else:
        count[word] = 1

print()
print("Word frequency:")
for word in sorted(count):
    print(" ", word, "=", count[word])

longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print()
print("Longest word:", longest)

file = open("paragraph_copy.txt", "w")
file.write(content)
file.close()
print("File copied to paragraph_copy.txt")
```

## Q16. Exception Handling in Banking Transactions

Uses try, except, else and finally, plus ValueError and ZeroDivisionError.

```python
def withdraw(balance, amount):
    if amount > balance:
        raise ValueError("Withdrawal amount is more than the balance")
    return balance - amount

try:
    balance = float(input("Enter account balance: "))
    amount = float(input("Enter withdrawal amount: "))
    transactions = int(input("Enter number of transactions today: "))

    average = 2500 / transactions
    new_balance = withdraw(balance, amount)

except ValueError as error:
    print("ValueError:", error)

except ZeroDivisionError:
    print("ZeroDivisionError: number of transactions cannot be zero")

else:
    print("Average transaction amount:", average)
    print("Withdrawal successful")
    print("Remaining balance:", new_balance)

finally:
    print("Transaction closed")
```

```python
# extra examples of each error
try:
    balance = float("eight thousand")
except ValueError:
    print("ValueError: invalid number")

try:
    print(2500 / 0)
except ZeroDivisionError:
    print("ZeroDivisionError: cannot divide by zero")

try:
    withdraw(5000, 7000)
except ValueError as error:
    print("ValueError:", error)
```

## Q19. Matrix Multiplication and Determinant Using NumPy

The inverse is printed only if the determinant is not zero.

```python
import numpy as np

print("Enter matrix A (3 x 3):")
a = []
for i in range(3):
    row = []
    for j in range(3):
        row.append(int(input("Enter element: ")))
    a.append(row)

print("Enter matrix B (3 x 3):")
b = []
for i in range(3):
    row = []
    for j in range(3):
        row.append(int(input("Enter element: ")))
    b.append(row)

A = np.array(a)
B = np.array(b)

det_a = round(np.linalg.det(A), 2)
det_b = round(np.linalg.det(B), 2)

print()
print("Matrix A:")
print(A)
print("Matrix B:")
print(B)
print("A x B (matrix multiplication):")
print(np.dot(A, B))
print("Determinant of A:", det_a)
print("Determinant of B:", det_b)

if det_a == 0:
    print("Inverse of A does not exist")
else:
    print("Inverse of A:")
    print(np.linalg.inv(A))

if det_b == 0:
    print("Inverse of B does not exist")
else:
    print("Inverse of B:")
    print(np.linalg.inv(B))

print("Transpose of A:")
print(A.T)
print("Transpose of B:")
print(B.T)
```

## Q20. Vector Operations Using NumPy

```python
import numpy as np

# Accept two vectors of 5 elements
a = np.array([int(input("Enter element of first vector: ")) for i in range(5)])
b = np.array([int(input("Enter element of second vector: ")) for i in range(5)])

# Vector addition
print("Addition:", a + b)

# Vector subtraction
print("Subtraction:", a - b)

# Element-wise multiplication
print("Multiplication:", a * b)

# Dot product
print("Dot product:", np.dot(a, b))

# Magnitude of vectors
mag_a = np.linalg.norm(a)
mag_b = np.linalg.norm(b)
print("Magnitude of first vector :", mag_a)
print("Magnitude of second vector:", mag_b)

# Angle between two vectors
cos_theta = np.dot(a, b) / (mag_a * mag_b)
angle = np.degrees(np.arccos(np.clip(cos_theta, -1, 1)))
print("Angle between vectors:", angle, "degrees")
```

## Q22. Student Dataset Preprocessing Using Pandas

Missing values are filled with the column mean and duplicate rows are removed.
The cleaned data is saved in `students_cleaned.csv`.

```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Ananya", "Rahul", "Sneha", "Arjun", "Meera",
             "Kiran", "Dev", "Priya", "Ananya"],
    "Roll No": [101, 102, 103, 104, 105, 106, 107, 108, 101],
    "Age": [20, 21, np.nan, 20, 22, 21, 19, np.nan, 20],
    "Marks": [78, 65, np.nan, 88, 92, 55, 70, 80, 78],
    "City": ["Kochi", "Kollam", "Kochi", np.nan,
             "Thrissur", "Kollam", "Kochi", "Thrissur", "Kochi"],
}

df = pd.DataFrame(data)

print("First five records:")
print(df.head())

print()
print("Missing values in each column:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna("Unknown")

print()
print("After filling missing values:")
print(df)

df = df.drop_duplicates()

print()
print("After removing duplicate records:")
print(df)

print()
print("Statistical information:")
print(df.describe())

df.to_csv("students_cleaned.csv", index=False)
print()
print("Cleaned data saved in students_cleaned.csv")
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Mouse", "Monitor", "Laptop", "Keyboard",
                "Monitor", "Mouse", "Laptop", "Keyboard"],
    "Quantity": [3, 25, 6, 2, 12, 4, 18, 1, 9],
    "Price": [55000, 550, 12500, 55000, 1200, 12500, 550, 55000, 1200],
    "Region": ["North", "South", "North", "East",
               "South", "East", "North", "South", "East"],
}

df = pd.DataFrame(data)
df["Sales"] = df["Quantity"] * df["Price"]

print("Sales records:")
print(df)

print()
print("Total sales of each product:")
print(df.groupby("Product")["Sales"].sum())

print()
print("Total revenue:", df["Sales"].sum())

print()
print("Products with sales above 10000:")
print(df[df["Sales"] > 10000][["Product", "Region", "Sales"]])

print()
print("Region wise revenue:")
print(df.groupby("Region")["Sales"].sum())

df.to_csv("sales.csv", index=False)
print()
print("Data saved in sales.csv")
```

## Q25. Integrated Python Data Processing System

Student report using Python functions, NumPy and Pandas. Invalid marks are
handled with try and except, and a blank mark is treated as a missing value.

```python
import numpy as np
import pandas as pd

subjects = ["English", "Maths", "Physics", "Computer", "Python"]

def total_marks(marks):
    total = 0
    for mark in marks:
        total = total + mark
    return total

def percentage(total):
    return total / 5

def grade(percent):
    if percent >= 90:
        return "A+"
    elif percent >= 80:
        return "A"
    elif percent >= 70:
        return "B+"
    elif percent >= 60:
        return "B"
    elif percent >= 50:
        return "C"
    elif percent >= 40:
        return "D"
    else:
        return "F"

students = []

for i in range(3):
    print("Student", i + 1)
    name = input("Enter name: ")
    roll = input("Enter roll number: ")

    marks = []
    for subject in subjects:
        value = input("Enter marks of " + subject + ": ")
        if value == "":
            marks.append(np.nan)
        else:
            try:
                mark = float(value)
                if mark < 0 or mark > 100:
                    print("Marks must be 0 to 100, taken as 0")
                    mark = 0
                marks.append(mark)
            except ValueError:
                print("Invalid marks, taken as 0")
                marks.append(0)

    students.append([roll, name, marks])

data = {"Roll No": [], "Name": []}
for subject in subjects:
    data[subject] = []

for student in students:
    data["Roll No"].append(student[0])
    data["Name"].append(student[1])
    for j in range(5):
        data[subjects[j]].append(student[2][j])

df = pd.DataFrame(data)

print()
print("Student data:")
print(df)

print()
print("Missing values:")
print(df.isnull().sum())

for subject in subjects:
    df[subject] = df[subject].fillna(df[subject].mean())

totals = []
percents = []
grades = []

for i in range(len(df)):
    marks = []
    for subject in subjects:
        marks.append(df.loc[i, subject])
    total = total_marks(marks)
    percent = percentage(total)
    totals.append(total)
    percents.append(round(percent, 2))
    grades.append(grade(percent))

df["Total"] = totals
df["Percentage"] = percents
df["Grade"] = grades

print()
print("Final report:")
print(df)

marks_array = df[subjects].values
print()
print("Average marks :", round(np.mean(marks_array), 2))
print("Maximum marks :", np.max(marks_array))
print("Minimum marks :", np.min(marks_array))

class_average = df["Total"].mean()
print()
print("Class average total:", round(class_average, 2))
print("Students above class average:")

for i in range(len(df)):
    if df.loc[i, "Total"] > class_average:
        print(" ", df.loc[i, "Name"], "-", df.loc[i, "Total"], "-", df.loc[i, "Grade"])

df.to_csv("student_report.csv", index=False)
print()
print("Report saved in student_report.csv")
```
